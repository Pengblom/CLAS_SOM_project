```python
# ==========================================
# LOAD CSV
# ==========================================

import pandas as pd
import numpy as np

FILE_PATH = "test_features_stroop/participant_53_stroop_windows.csv" 

df = pd.read_csv(FILE_PATH)

print("CSV loaded successfully")
```

```python
# ==========================================
# BASIC STRUCTURE
# ==========================================

print("Shape (rows, columns):", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
print(df.head())
```

```python
# ==========================================
# STATISTICS
# ==========================================

print("Descriptive statistics:")
print(df.describe())
```

```python
# ==========================================
# EXTRA STATS
# ==========================================

print("Min values:")
print(df.min(numeric_only=True))

print("\nMax values:")
print(df.max(numeric_only=True))

print("\nStd values:")
print(df.std(numeric_only=True))
```

```python
# ==========================================
# DATA QUALITY CHECK
# ==========================================

print("Missing values per column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
# ==========================================
# SELECT SPECIFIC FEATURE COLUMNS
# ==========================================

columns = [
    "HR",
    "HRV_RMSSD",
    "HRV_SDNN",
    "SCR_Count",
    "EDA_Tonic_log",
    "EDA_Phasic_log"
]

# säkerställ att alla finns
missing = [c for c in columns if c not in df.columns]
if missing:
    raise ValueError(f"Saknade kolumner: {missing}")

features = df[columns].values.astype(np.float32)

print("Shape:", features.shape)  # ska vara (202, 6)
print("First row:", features[0])
```

```python
# ==========================================
# EXPORT TO C ARRAY (2D)
# ==========================================

def export_features_to_c(features, name="input_data"):

    rows, cols = features.shape

    print(f"#define NUM_SAMPLES {rows}")
    print(f"#define NUM_FEATURES {cols}\n")

    print(f"static const float {name}[{rows}][{cols}] = {{")

    for r in range(rows):
        print("   {", end="")

        for c in range(cols):
            print(f"{features[r, c]:.6f}f", end="")
            if c < cols - 1:
                print(", ", end="")

        print("},")

    print("};")


export_features_to_c(features)
```
