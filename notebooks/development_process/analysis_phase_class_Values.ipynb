```python
# ======================================
# COMPLETE PARTICIPANT ANALYSIS
# ======================================

import pandas as pd

# ======================================
# LOAD DATA
# ======================================

file_path = "../Data/stresstest2/training_data_improved_5.csv"

df = pd.read_csv(file_path, sep=';')

# ======================================
# CONVERT FEATURES TO NUMERIC
# ======================================

features = ["HR", "RMSSD", "SC_PH", "SC_RR"]

for feature in features:

    df[feature] = pd.to_numeric(
        df[feature],
        errors="coerce"
    )

# ======================================
# CHECK INVALID VALUES
# ======================================

print("\n===== INVALID VALUES =====")

for feature in features:

    invalid_count = df[feature].isna().sum()

    print(f"{feature}: {invalid_count} invalid values")
# ======================================
# SETTINGS
# ======================================

print("Participant = 2 ")

classification_col = "baseline_classification"
# classification_col = "improved_classification"


features = ["HR", "RMSSD", "SC_PH", "SC_RR"]


# ======================================
# PHASE GROUPS
# ======================================

rest_phases = [2,4,6,8,10,12]
stroop_phases = [1,3,5]
mat_phases = [7,9,11]

# ======================================
# PHASE FUNCTION
# ======================================

def summarize_phase(dataframe, phases, phase_name):

    phase_df = dataframe[
        dataframe["Experiment_Phase"].isin(phases)
    ]

    total = len(phase_df)

    print(f"\n========== {phase_name} ==========")
    print(f"Total classifications: {total}\n")

    for c in [0,1,2]:

        count = (
            phase_df[classification_col] == c
        ).sum()

        percentage = (count / total) * 100

        print(
            f"Class {c}: "
            f"{count} classifications "
            f"({percentage:.2f}%)"
        )

# ======================================
# RUN PHASE ANALYSIS
# ======================================

print("\n======================================")
print("PHASE CLASSIFICATION ANALYSIS")
print("======================================")

summarize_phase(df, rest_phases, "REST")
summarize_phase(df, stroop_phases, "STROOP")
summarize_phase(df, mat_phases, "MAT")

# ======================================
# PHYSIOLOGICAL ANALYSIS
# ======================================

print("\n======================================")
print("PHYSIOLOGICAL ANALYSIS")
print("======================================")

# ======================================
# BASELINE
# ======================================

print("\n===== BASELINE =====")

baseline_total = len(df)

print(f"Total classifications: {baseline_total}\n")

for feature in features:

    mean_value = df[feature].mean()

    print(f"{feature}: {mean_value:.2f}")

# ======================================
# CLASS ANALYSIS
# ======================================

for c in [0,1,2]:

    class_df = df[
        df[classification_col] == c
    ]

    total_classifications = len(class_df)

    print(f"\n===== CLASS {c} =====")
    print(f"Total classifications: {total_classifications}\n")

    for feature in features:

        mean_value = class_df[feature].mean()

        print(f"{feature}: {mean_value:.2f}")
```
