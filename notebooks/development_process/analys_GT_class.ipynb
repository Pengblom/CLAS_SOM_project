```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.spatial.distance import euclidean
from sklearn.metrics import confusion_matrix
```

```python
# ==========================================
# LOAD DATA
# ==========================================

#file_path = "../Data/stresstest1_train/training_data8.csv"

# test 2 
file_path = "../Data/stresstest2/training_data_improved_8.csv"

df = pd.read_csv(file_path, sep=';')

print(df.head())

print("\nColumns:")
print(df.columns)
```

```python
# ==========================================
# FULL CLASSIFICATION EVALUATION
# - Remove class 255
# - Confusion Matrix
# - Precision
# - Recall
# - F1-score
# - Support
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

print("PARTICIPANT 8")
print("TEST 2")

# ==========================================
# REMOVE INVALID CLASS 255
# ==========================================

filtered_df = df[
    df["baseline_classification"] != 255
].copy()

# ==========================================
# TRUE + PREDICTED LABELS
# ==========================================

y_true = filtered_df["ground_truth"]
y_pred = filtered_df["baseline_classification"]

# ==========================================
# ACCURACY
# ==========================================

accuracy = accuracy_score(y_true, y_pred)

print("===================================================")
print(f"OVERALL ACCURACY: {accuracy*100:.2f}%")
print("===================================================")

# ==========================================
# CLASSIFICATION REPORT
# ==========================================

report = classification_report(
    y_true,
    y_pred,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

print("\n===================================================")
print("CLASSIFICATION REPORT")
print("===================================================")

print(report_df.round(3))

# ==========================================
# CONFUSION MATRIX
# ==========================================

cm = confusion_matrix(y_true, y_pred)

# ------------------------------------------
# Normalize row-wise
# ------------------------------------------

row_sums = cm.sum(axis=1, keepdims=True)

cm_percent = np.divide(
    cm.astype("float"),
    row_sums,
    where=row_sums != 0
) * 100

# ==========================================
# PLOT
# ==========================================

fig, ax = plt.subplots(figsize=(9, 6))

im = ax.imshow(cm_percent)

# ------------------------------------------
# Labels
# ------------------------------------------

classes = np.unique(y_true)

ax.set_xticks(np.arange(len(classes)))
ax.set_yticks(np.arange(len(classes)))

ax.set_xticklabels(classes, fontsize=12)
ax.set_yticklabels(classes, fontsize=12)

ax.set_xlabel("Predicted SOM Class", fontsize=13)
ax.set_ylabel("Ground Truth", fontsize=13)

ax.set_title(
    f"Participant 8 - Confusion Matrix Test2 (%)\nAccuracy = {accuracy*100:.2f}%",
    fontsize=16,
    pad=20
)

# ------------------------------------------
# Text inside cells
# ------------------------------------------

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):

        count = cm[i, j]
        percent = cm_percent[i, j]

        # Handle NaN values
        if np.isnan(percent):
            text = f"{count}\n(N/A)"
            text_color = "black"
        else:
            text = f"{count}\n({percent:.1f}%)"
            text_color = "white" if percent > 50 else "black"

        ax.text(
            j,
            i,
            text,
            ha="center",
            va="center",
            fontsize=11,
            fontweight="bold",
            color=text_color
        )
# ------------------------------------------
# Colorbar
# ------------------------------------------

cbar = plt.colorbar(im)
cbar.set_label("Row Percentage (%)", fontsize=12)

# ------------------------------------------
# Grid
# ------------------------------------------

ax.set_xticks(np.arange(-.5, len(classes), 1), minor=True)
ax.set_yticks(np.arange(-.5, len(classes), 1), minor=True)

ax.grid(which="minor", color="white", linestyle="-", linewidth=2)

ax.tick_params(which="minor", bottom=False, left=False)

plt.tight_layout()
plt.show()
```
