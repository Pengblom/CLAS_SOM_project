```python
import pandas as pd
import matplotlib.pyplot as plt

rows = []

with open("data.csv", "r", encoding="latin1") as f:
    for line in f:
        line = line.strip()
        
        # 🔹 ta bara rader som börjar med DATA:
        if line.startswith("DATA:"):
            line = line.replace("DATA:", "")
            rows.append(line)

# 🔹 skapa DataFrame direkt
df = pd.DataFrame(
    [r.split(",") for r in rows],
    columns=["time_ms", "sc_ph", "sc_rr", "scr_count"]
)

# 🔹 konvertera typer
df = df.astype({
    "time_ms": float,
    "sc_ph": float,
    "sc_rr": float,
    "scr_count": int
})

df["time_s"] = df["time_ms"] / 1000

print("Rows loaded:", len(df))
print(df.head())
print(df.describe())

# 🔹 plot
plt.figure(figsize=(12,6))
plt.plot(df["time_s"], df["sc_rr"], label="SC_RR")
plt.plot(df["time_s"], df["sc_ph"], label="SC_PH")

plt.xlabel("Time (s)")
plt.ylabel("Value")
plt.title("EDA Features")
plt.legend()
plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

rows = []

with open("data4.csv", "r", encoding="utf-16") as f:
    for line in f:
        line = line.strip()
        
        if line.startswith("DATA:"):
            line = line.replace("DATA:", "")
            rows.append(line)

if len(rows) == 0:
    print("❌ No DATA rows found")
else:
    df = pd.DataFrame(
        [r.split(",") for r in rows],
        columns=["time_ms", "sc_ph", "sc_rr", "scr_count"]
    )

    df = df.astype({
        "time_ms": float,
        "sc_ph": float,
        "sc_rr": float,
        "scr_count": int
    })

    df["time_s"] = df["time_ms"] / 1000

    print("✅ Rows loaded:", len(df))
    print(df.head())

    plt.figure(figsize=(12,6))
    plt.plot(df["time_s"], df["sc_rr"], label="SC_RR")
    plt.plot(df["time_s"], df["sc_ph"], label="SC_PH")
    plt.legend()
    plt.grid()
    plt.show()
```
