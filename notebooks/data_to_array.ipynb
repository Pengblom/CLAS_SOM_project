```python
import csv
from pathlib import Path

# === Inställningar ===
data_file = Path("../data/raw/Participants/Part12/full_gsr_ppg.csv")

start_sec = 1200
end_sec = 1260

subset = []
t0 = None  # starttid

# === Läs och filtrera data ===
with open(data_file) as f:
    reader = csv.reader(f)
    next(reader)  # hoppa över header
    
    for row in reader:
        if len(row) > 6:
            try:
                timestamp = float(row[1])   # PythonTimestamp
                value = float(row[6])       # GSR
                
                if t0 is None:
                    t0 = timestamp  # sätt första giltiga tid
                
                t = timestamp - t0  # relativ tid i sekunder
                
                if start_sec <= t <= end_sec:
                    subset.append(value)
                    
            except:
                continue  # skippar skräprader

# === Kontroll ===
print("Antal datapunkter:", len(subset))

# === Spara till C-header ===
with open("data_subset.h", "w") as f:
    f.write(f"const float sensor_data[{len(subset)}] = {{\n")
    f.write(",\n".join(map(str, subset)))
    f.write("\n};")

print("Klart! Fil sparad som data_subset.h")
```
