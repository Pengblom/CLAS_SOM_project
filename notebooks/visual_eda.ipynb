```python
import pandas as pd
import matplotlib.pyplot as plt

file = "../data/raw/Participants/Part10/full_gsr_ppg.csv"

data = pd.read_csv(file)

# ta bort rader där Timestamp inte är numerisk
data = data[pd.to_numeric(data["Timestamp"], errors="coerce").notnull()]

# konvertera till nummer
data["PythonTimestamp"] = data["PythonTimestamp"].astype(float)
data["gsr"] = data["gsr"].astype(float)

# gör relativ tid
time = data["PythonTimestamp"] - data["PythonTimestamp"].iloc[0]
gsr = data["gsr"]

plt.figure(figsize=(12,5))
plt.plot(time, gsr)

plt.title("CLAS EDA signal")
plt.xlabel("Time (seconds)")
plt.ylabel("GSR / EDA")
plt.grid(True)

plt.show()
```
